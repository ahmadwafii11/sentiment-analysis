# 1. Integration Data File Emotion Tweet to Sentiment Tweet

```python
dataset_tweet_emotion
```

```python
# rename column

dataset_tweet_emotion = dataset_tweet_emotion[['tweet','label']].rename(columns={'tweet':'Tweet'})
dataset_tweet_emotion
```

```python
# mapping emotion to sentiment

emotion_to_sentiment = {
    'happy': 'positive',
    'love': 'positive',
    'anger': 'negative',
    'sadness': 'negative',
    'fear': 'negative'
}

# change label emotion to sentiment
dataset_tweet_emotion['label'] = dataset_tweet_emotion['label'].map(emotion_to_sentiment)
dataset_tweet_emotion
```

# 2. Integration Data File Tweet Banjir Sumatra

```python
dataset_keyword_banjir
```

```python
# rename column chat and rate

dataset_keyword_banjir = dataset_keyword_banjir[['chat','rate']].rename(columns={'chat':'Tweet',
                                                                             'rate' : 'label'})
dataset_keyword_banjir
```

```python
# mapping lowercase label

rename_lowecase_label = {
    'Positif': 'positive',
    'Netral': 'neutral',
    'Negatif': 'negative',
}

# rename lowercase label
dataset_keyword_banjir['label'] = dataset_keyword_banjir['label'].map(rename_lowecase_label)
dataset_keyword_banjir
```

# 3. Integration Data File Tweet Keyword Jokowi

```python
dataset_keyword_jokowi
```

```python
# rename column text and label

dataset_keyword_jokowi = dataset_keyword_jokowi[['text','Label']].rename(columns={'text':'Tweet',
                                                                             'Label' : 'label'})
dataset_keyword_jokowi
```

```python
# count label

dataset_keyword_jokowi['label'].value_counts()
```

```python
# count label

dataset_keyword_jokowi['label'].value_counts()
```

# 5. Integration Data File Tweet PPKM from Sentiment Label Number to Text

```python
dataset_ppkm
```

```python
# drop coloumn date and user

dataset_ppkm.drop(columns=["Date", "User"], inplace=True)

# rename column sentiment to label
dataset_ppkm.rename(columns={"sentiment": "label"}, inplace=True)

dataset_ppkm
```

```python
# mapping number label sentiment to text label sentiment

number_to_sentiment = {
    0: 'positive',
    1: 'neutral',
    2: 'negative'
}

# change number label sentiment to text label sentiment

dataset_ppkm['label'] = dataset_ppkm['label'].map(number_to_sentiment)
dataset_ppkm
```

# 6. Integration Data File Tweet Label Sentiment ITE

```python
dataset_label_ite
```

```python
# rename column 

dataset_label_ite = dataset_label_ite[['Tweet','sentimen']].rename(columns={'Tweet':'Tweet',
                                                                             'sentimen' : 'label'})
dataset_label_ite
```

```python
# count label
dataset_label_ite['label'].value_counts()
```

```python
# mapping number score label sentiment to text label sentiment

number_to_sentiment_ite = {
    0.0: 'neutral',
    2.0: 'negative',
    1.0: 'positive',
}

# change number score label sentiment to text label sentiment

dataset_label_ite['label'] = dataset_label_ite['label'].map(number_to_sentiment_ite)
dataset_label_ite
```

```python
# check missing value

dataset_label_ite.isnull().sum()
```

```python
# delete row missing value (NaN)
dataset_label_ite = dataset_label_ite.dropna(subset=['Tweet', 'label'])

# Kadang data tidak NaN, tapi isinya cuma spasi setelah dibersihkan
dataset_label_ite = dataset_label_ite[dataset_label_ite['label'].str.strip() != '']

# Agar urutan baris kembali rapi (0, 1, 2, ...) tanpa ada angka yang melompat
dataset_label_ite = dataset_label_ite.reset_index(drop=True)

dataset_label_ite
```

```python
# count label
dataset_label_ite['label'].value_counts()
```

```python
# check missing value

dataset_label_ite.isnull().sum()
```

# Merge DataFrame 1, 2, 3, 4, 5

```python
# merge dataframe 

dataset_training = pd.concat([dataset_tweet_emotion, dataset_keyword_banjir, dataset_keyword_jokowi, dataset_pilpres, dataset_ppkm, 
                              dataset_label_ite, dataset_youtube_comment, dataset_sent_net_indo, dataset_review_startup], ignore_index=True)
dataset_training
```

```python
# count label
dataset_training['label'].value_counts()
```

```python
# check missing value

dataset_training.isnull().sum()
```

```python
# delete row missing value (NaN)
dataset_training = dataset_training.dropna(subset=['Tweet', 'label'])

# Kadang data tidak NaN, tapi isinya cuma spasi setelah dibersihkan
dataset_training = dataset_training[dataset_training['Tweet'].str.strip() != '']

# Agar urutan baris kembali rapi (0, 1, 2, ...) tanpa ada angka yang melompat
dataset_training = dataset_training.reset_index(drop=True)

dataset_training
```
